```python
# importing requried libraries

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings("ignore")
```

```python
# loading dataset


df = pd.read_csv("creditcard.csv")

df.head()
```

```python
#Inspect Dataset Structure

# Dataset Dimensions

df.shape
```

The dataset contains 284,807 transactions and 31 attributes.

```python
# displaying column information

df.info()
```

```python
# statistical summary

df.describe()
```

```python
# data cleaning
# checking missing values

df.isnull().sum()
```

The dataset does not contain missing values. Therefore, no imputation technique is required.

```python
# Checking Duplicate Records
df.duplicated().sum()
```

```python
# removing duplicate transactions

df = df.drop_duplicates()

df.shape
```

Removing duplicate transactions prevents incorrect fraud patterns from influencing future models.

```python
# Checking Data Types
df.dtypes
```

All attributes are numerical, which makes the dataset suitable for machine learning algorithms without requiring categorical encoding.

```python
# Handling Class Imbalance
```

```python
# Fraud datasets usually contain significantly fewer fraudulent transactions compared with normal transactions.

#Check class distribution:

df["Class"].value_counts()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x="Class",
    data=df
)

plt.title("Distribution of Fraud and Normal Transactions")

plt.xlabel("Class")
plt.ylabel("Number of Transactions")

plt.show()
```

Observation:

The dataset is highly imbalanced:

* Class 0 → Normal transactions
* Class 1 → Fraudulent transactions

Only a very small percentage of transactions are fraudulent.

This imbalance will need to be addressed in future modeling using techniques such as:

* SMOTE oversampling
* Undersampling
* Class weighting

```python
# Exploratory Data Analysis (EDA)

## Transaction Amount Distribution

plt.figure(figsize=(8,5))

sns.histplot(

    df["Amount"],

    bins=50,

    kde=True

)

plt.title("Transaction Amount Distribution")

plt.xlabel("Amount ($)")

plt.show()
```

Insight:

Most transactions involve smaller amounts, while a small number of transactions contain very high amounts.


```python
# Fraud vs Transaction Amount
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Class",
    y="Amount",
    data=df
)

plt.title(
    "Transaction Amount by Class"
)

plt.show()
```

Insight:

Fraudulent transactions show different spending patterns compared with normal transactions. Transaction amount can therefore be an important feature.

```python
# Correlation Analysis

plt.figure(figsize=(12,8))

correlation = df.corr()

sns.heatmap(
    correlation,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Feature Correlation Matrix")

plt.show()
```

Insight:

The PCA-transformed variables (V1-V28) have different relationships with the target variable. Some features show stronger correlations with fraudulent transactions.

```python
# Fraud Transaction Analysis

fraud = df[df["Class"] == 1]

normal = df[df["Class"] == 0]


print(
    "Normal Transactions:",
    len(normal)
)

print(
    "Fraud Transactions:",
    len(fraud)
)
```

Insight:

The number of legitimate transactions is much higher than fraudulent transactions. This confirms the need for imbalance handling during model development.

```python
# Time Analysis

plt.figure(figsize=(8,5))

sns.histplot(
    fraud["Time"],
    bins=50
)

plt.title(
    "Time Distribution of Fraud Transactions"
)

plt.xlabel("Time")

plt.show()
```

Insight:

Fraudulent transactions are not evenly distributed throughout the transaction period, suggesting that time-related patterns may help improve detection models.

Feature Engineering

Additional features will be created for future modeling.

```python
# Transaction Hour

df["Hour"] = (

    df["Time"] // 3600

) % 24
```

```python
# Amount Categories

df["Amount_Category"] = pd.cut(

    df["Amount"],

    bins=[

        -1,

        50,

        200,

        1000,

        float("inf")

    ],

    labels=[

        "Low",

        "Medium",

        "High",

        "Very High"

    ]

)
```
